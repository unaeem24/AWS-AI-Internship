```python
#libraries for data cleaning
import pandas as pd
import numpy as np
```

```python
churn_data = pd.read_csv('Telco-Customer-Churn.csv')
churn_data_raw = churn_data.copy()
```

```python
churn_data.dtypes
```

```python
churn_data
```

```python
#Convert spaces to nan (Not a Number) for better handling of missing values

churn_data['TotalCharges'] = churn_data['TotalCharges'].replace(' ', np.nan)
# Convert TotalCharges to numeric, coercing errors to NaN
churn_data['TotalCharges'] = pd.to_numeric(churn_data['TotalCharges'], errors='coerce')
```

```python
churn_data
```

```python
churn_data.isnull().sum()
```

```python
churn_data['TotalCharges'] = churn_data['TotalCharges'].fillna(0)
churn_data.isnull().sum()
```

```python
churn_data.drop('customerID', axis=1, inplace=True)
```

```python
# This will automatically convert all text columns into 0s and 1s
churn_data = pd.get_dummies(churn_data, drop_first=True)
```

```python
churn_data.dtypes
```

```python
churn_data.columns
```

```python
from sklearn.preprocessing import StandardScaler

# 1. Initialize the scaler
scaler = StandardScaler()

# 2. List the columns that need scaling (leave the 0/1 columns alone!)
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

# 3. Fit and transform the training data
churn_data[num_cols] = scaler.fit_transform(churn_data[num_cols])

# 4. Transform the test data (using the same parameters from training)
churn_data[num_cols] = scaler.transform(churn_data[num_cols])
```

```python
#save the data in csv file
churn_data.to_csv('churn_data_cleaned.csv', index=False)
```
