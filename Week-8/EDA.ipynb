```python
#Libraries for data manipulation and machine learning
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns


```

```python
churn_data = pd.read_csv('Telco-Customer-Churn.csv')
churn_data_raw = pd.read_csv('Telco-Customer-Churn.csv')
```

```python
churn_data.isnull().sum()
```

```python
# pi chart for churn distribution
plt.figure(figsize=(6, 6))
churn_counts = churn_data['Churn'].value_counts()
plt.pie(churn_counts, labels=churn_counts.index, autopct='%1.1f%%')
plt.title('Churn Distribution')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# Create the grouped bar chart
sns.countplot(data=churn_data, x='Contract', hue='Churn', palette='viridis')

# Adding labels and title
plt.title('Churn Distribution by Contract Type', fontsize=15)
plt.xlabel('Contract Type', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.legend(title='Churned?', labels=['No', 'Yes'])

plt.show()
```

```python
plt.figure(figsize=(10, 6))

# KDE Plot to see the density of churn across monthly charges
sns.kdeplot(data=churn_data[churn_data['Churn'] == 'No']['MonthlyCharges'], 
            label='Not Churned', shade=True, color='blue')
sns.kdeplot(data=churn_data[churn_data['Churn'] == 'Yes']['MonthlyCharges'], 
            label='Churned', shade=True, color='red')

plt.title('Distribution of Monthly Charges by Churn Status', fontsize=15)
plt.xlabel('Monthly Charges ($)', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.legend()
plt.show()
```
